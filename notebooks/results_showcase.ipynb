```python
import sys; sys.path.append("..")
import torch
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

from src.config import get_guitarset_dir
from src.data.guitarset import GuitarSetDataset, load_or_create_split
from src.model import UNetPianoRoll
from src.losses import CombinedLoss
from src.training import train
from src.evaluation import inference, get_metrics, plot_pianoroll_comparison

DATA_DIR = get_guitarset_dir()  # set the GUITARSET_DIR env var to your GuitarSet root
```

# Training

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = UNetPianoRoll(dropout_rate=0.5).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
```

```python
dataset = GuitarSetDataset(DATA_DIR)

# persistent train/test split, stored as portable basenames in ../test_split.json
train_samples, test_samples = load_or_create_split(dataset.samples, split_path="../test_split.json")

# Initialize datasets
train_dataset = GuitarSetDataset(DATA_DIR, apply_nmf=False, apply_sections=True)
train_dataset.samples = train_samples

test_sections_dataset = GuitarSetDataset(DATA_DIR, apply_nmf=False, apply_sections=True)
test_sections_dataset.samples = test_samples

test_fullroll_dataset = GuitarSetDataset(DATA_DIR, apply_nmf=False, apply_sections=False)
test_fullroll_dataset.samples = test_samples

# Create the training data loader (sections of one track per step)
train_loader = DataLoader(
    train_dataset, batch_size=1, shuffle=True, num_workers=8,
    pin_memory=True, persistent_workers=True, prefetch_factor=8
)
```

```python
# BCE + MSE combined loss (see src/losses.py)
criterion = CombinedLoss(alpha=0.5).to(device)
```

```python
# training loop in src/training.py: gradient accumulation over the sections of each
# track, one optimizer step per track, checkpoint saved on epoch-loss improvement
```

```python
losses, first_last_losses = train(model, train_loader, criterion, optimizer, num_epochs=144, checkpoint_dir="../models")
```

# Evaluation
